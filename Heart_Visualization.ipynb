```python
import seaborn as sns
```

```python
df=pd.read_csv("heart .csv")
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True),cmap="coolwarm",annot=True)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.lineplot(data=df.sort_values(by="age"),x='age',y='thalachh',marker='o',color='blue')
plt.title("Max Heart Rate by Age")
plt.xlabel('Age')
plt.ylabel('Max Heart Rate')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.lineplot(data=df.sort_values(by="age"),x="age",y='chol',marker='o',color='red')
plt.title("Cholestrol Level By Age")
plt.xlabel("Age")
plt.ylabel("Cholestrol")
plt.show()
```

```python
plt.figure(figsize=(18,10))
sns.barplot(data=df,x='cp',y='chol',palette="coolwarm")
plt.title("Chest Pain by Cholestrol Level")
plt.xlabel("Chest Pain Level")
plt.ylabel("Cholestrol Level")
```

```python
# Heart disease count
sns.countplot(x='output', data=df)
plt.title("Heart Disease Count")
plt.xlabel("Heart Disease (1 = Yes, 0 = No)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(data=df,x="cp")
plt.title("Chest Pain Type Count")
plt.xlabel("Chest Pain Type")
plt.ylabel("Count ")
plt.show()
```

```python
sns.scatterplot(x='age', y='thalachh', hue='output', data=df)
plt.title("Age vs Max Heart Rate by Heart Disease")
plt.xlabel("Age")
plt.ylabel("Max Heart Rate (thalachh)")
plt.show()
```
